# 01 — Data Inventory & Validation

## 1. Objective

This notebook validates the raw Olist Brazilian E-Commerce dataset before
analytical processing for the customer retention capstone.

It establishes:
- dataset completeness and structure
- key and relationship integrity
- table cardinality and analytical grain
- data-quality issues requiring documented treatment
- feasibility of the proposed retention analysis

No analytical records are removed or transformed without a documented
treatment decision. The final customer-level analytical dataset will be
constructed only after these validation checks are complete.

The notebook is designed to provide a reproducible evidence trail from
raw source data to a validated analytical starting point.

The project uses a fixed local project structure so that raw data,
notebooks and analytical outputs remain separated.

## 5. Key & Relationship Validation

This section validates candidate keys and relationships between the
nine raw tables before any analytical joins are performed.

The purpose is to identify duplicate keys, unexpected one-to-many
relationships and orphan records that could create incorrect row
multiplication or biased analytical results.

No data is modified at this stage.

### 5.1 Candidate Key Validation

Candidate keys are checked for uniqueness within their respective source
tables.

A field is treated as a valid table-level key only when the number of
unique values equals the number of rows.

```python
# Candidate key uniqueness

candidate_keys = {
    "olist_orders_dataset.csv": ["order_id"],
    "olist_customers_dataset.csv": ["customer_id"],
    "olist_products_dataset.csv": ["product_id"],
    "olist_sellers_dataset.csv": ["seller_id"],
    "olist_order_items_dataset.csv": ["order_id", "order_item_id"],
    "olist_order_payments_dataset.csv": ["order_id", "payment_sequential"],
    "olist_order_reviews_dataset.csv": ["review_id"],
    "product_category_name_translation.csv": ["product_category_name"],
}

key_results = []

for filename, keys in candidate_keys.items():
    df = data[filename]

    for key in keys:
        duplicate_count = df[key].duplicated().sum()
        unique_count = df[key].nunique()

        key_results.append({
            "file": filename,
            "candidate_key": key,
            "unique_values": unique_count,
            "rows": len(df),
            "duplicate_rows": duplicate_count,
            "unique": duplicate_count == 0
        })

key_validation = pd.DataFrame(key_results)

display(key_validation)
```

This validation establishes the structural keys required for safe
relationships between the source tables. Customer identity is handled
separately because `customer_unique_id` represents the underlying
customer and is intentionally not expected to be unique in the customer
table.

```python
# Customer identity is not the same as the customer table key.
# customer_unique_id represents the underlying customer and may occur
# across multiple customer_id records.

customers = data["olist_customers_dataset.csv"]

customer_identity_summary = pd.DataFrame({
    "metric": [
        "Customer table rows",
        "Unique customer_id",
        "Unique customer_unique_id",
        "customer_unique_id duplicate rows"
    ],
    "value": [
        len(customers),
        customers["customer_id"].nunique(),
        customers["customer_unique_id"].nunique(),
        customers["customer_unique_id"].duplicated().sum()
    ]
})

display(customer_identity_summary)
```

### 5.2 Referential Integrity Validation

Foreign-key relationships are checked to determine whether every
non-null child key has a corresponding parent key.

This validates that the core source tables can be joined without
unmatched foreign-key records.


```python
# Foreign-key coverage

relationship_checks = []

def check_fk(child_file, child_column, parent_file, parent_column):
    child = data[child_file]
    parent = data[parent_file]

    parent_values = set(parent[parent_column].dropna().unique())
    child_values = child[child_column].dropna()

    matched = child_values.isin(parent_values)

    return {
        "child_table": child_file,
        "child_column": child_column,
        "parent_table": parent_file,
        "parent_column": parent_column,
        "child_non_null": len(child_values),
        "matched": matched.sum(),
        "orphan_rows": (~matched).sum(),
        "orphan_pct": round((~matched).mean() * 100, 4)
    }


relationships = [
    (
        "olist_orders_dataset.csv",
        "customer_id",
        "olist_customers_dataset.csv",
        "customer_id"
    ),
    (
        "olist_order_items_dataset.csv",
        "order_id",
        "olist_orders_dataset.csv",
        "order_id"
    ),
    (
        "olist_order_items_dataset.csv",
        "product_id",
        "olist_products_dataset.csv",
        "product_id"
    ),
    (
        "olist_order_items_dataset.csv",
        "seller_id",
        "olist_sellers_dataset.csv",
        "seller_id"
    ),
    (
        "olist_order_reviews_dataset.csv",
        "order_id",
        "olist_orders_dataset.csv",
        "order_id"
    ),
    (
        "olist_order_payments_dataset.csv",
        "order_id",
        "olist_orders_dataset.csv",
        "order_id"
    ),
]

relationship_validation = pd.DataFrame([
    check_fk(*relationship)
    for relationship in relationships
])

display(relationship_validation)
```

### Referential Integrity Result

All tested foreign-key relationships have complete parent coverage, with
zero orphan rows identified.

The validated relationships are:

- orders → customers
- order items → orders
- order items → products
- order items → sellers
- reviews → orders
- payments → orders

These relationships support the joins required for constructing the
order-level analytical dataset.

## 6. Relationship Cardinality & Key Structure

The source tables contain several one-to-many relationships. These
cardinalities are validated before analytical joins are performed to
prevent row multiplication and incorrect aggregation.

The checks focus on order items, payments and reviews, and on the
distinction between customer table identity and underlying customer
identity.

These findings determine the correct aggregation grain for the
analytical dataset.

### 6.1 Order → Item Cardinality

An order may contain one or more order-item records. The number of items
per order is therefore measured before item-level data is aggregated to
the order grain.

This check establishes the extent of one-to-many expansion and identifies
whether any orders contain unusually large numbers of items.

```python
order_item_counts = (
    data["olist_order_items_dataset.csv"]
    .groupby("order_id")
    .size()
    .rename("item_count")
)

print("Order item-count distribution:")
display(order_item_counts.describe())

print(
    f"Orders with multiple items: "
    f"{(order_item_counts > 1).sum():,}"
)

print(
    f"Maximum items in one order: "
    f"{order_item_counts.max():,}"
)
```

### Order → Item Cardinality Result

The order-items table contains 98,666 distinct orders, with an average of
1.14 items per order.

A total of 9,803 orders contain multiple item records, with a maximum of
21 items in a single order.

Therefore, order-item data must be aggregated to the order level before
being joined with other order-level measures. This prevents item-level
row multiplication from distorting order value and freight calculations.

### 6.2 Order → Payment Cardinality

An order can contain multiple payment records. Payment-level data must
therefore be aggregated to the order level before being joined with other
order-level data.

This check quantifies the one-to-many relationship and identifies orders
with unusually high numbers of payment records.

```python
payment_counts = (
    data["olist_order_payments_dataset.csv"]
    .groupby("order_id")
    .size()
    .rename("payment_count")
)

print("Payment-count distribution:")
display(payment_counts.describe())

print(
    f"Orders with multiple payment records: "
    f"{(payment_counts > 1).sum():,}"
)

print(
    f"Maximum payments in one order: "
    f"{payment_counts.max():,}"
)
```

### Order → Payment Cardinality Result

The payment table contains 99,440 distinct orders, with an average of
1.04 payment records per order.

A total of 2,961 orders contain multiple payment records, with a maximum
of 29 payment records for a single order.

Therefore, payment records must be aggregated to the order level before
being joined with other order-level measures. This prevents payment-level
row multiplication from affecting downstream order and customer metrics.

### 6.3 Order → Review Cardinality

An order may have more than one review record in the raw review table.
This cardinality is measured before reviews are incorporated into the
order-level analytical dataset.

The check determines whether review scores need to be aggregated to one
order-level value to avoid row multiplication.

```python
review_counts = (
    data["olist_order_reviews_dataset.csv"]
    .groupby("order_id")
    .size()
    .rename("review_count")
)

print("Orders with reviews:", f"{review_counts.shape[0]:,}")

print(
    "Orders with multiple reviews:",
    f"{(review_counts > 1).sum():,}"
)

print(
    "Maximum reviews for one order:",
    f"{review_counts.max():,}"
)
```

### Order → Review Cardinality Result

A total of 98,673 orders have at least one review record.

Multiple review records occur for 547 orders, with a maximum of three
reviews for a single order.

Therefore, review data must be reduced to the order level before being
joined with the analytical order table. Review scores will be aggregated
at order level rather than allowing multiple review rows to multiply
order-level records.

### 6.4 Review Key Structure

The review table contains a `review_id` field, but its uniqueness is
validated separately from the order-to-review relationship.

This check examines duplicated review IDs and determines whether the
duplicated identifiers are associated with multiple orders or different
review scores.

The result informs the appropriate review-level aggregation strategy and
prevents `review_id` from being incorrectly treated as a unique analytical
key.

```python
## Review key structure

reviews = data["olist_order_reviews_dataset.csv"]

# Review IDs are not unique in the raw dataset.
review_id_stats = (
    reviews.groupby("review_id")
    .agg(
        row_count=("review_id", "size"),
        order_count=("order_id", "nunique"),
        score_count=("review_score", "nunique")
    )
)

print(f"Total review rows: {len(reviews):,}")
print(f"Unique review IDs: {reviews['review_id'].nunique():,}")
print(
    f"Review IDs appearing more than once: "
    f"{(review_id_stats['row_count'] > 1).sum():,}"
)

print("\nReview ID duplication pattern:")
display(
    review_id_stats[review_id_stats["row_count"] > 1]
    .describe()
)
```

### Review Key Structure Result

The review table contains 99,224 rows and 98,410 unique `review_id` values.
A total of 789 review IDs appear more than once.

Each duplicated review ID occurs in two or three rows and is associated
with two or three orders. The review score remains consistent within
each duplicated review ID.

Therefore, `review_id` is not treated as a unique analytical key.
Reviews are related to the analytical dataset through `order_id`, with
review information subsequently aggregated to the order level.

### 6.5 Customer Identity Cardinality

The customer table contains a unique `customer_id` for each customer
record, but multiple customer records can map to the same underlying
customer through `customer_unique_id`.

The distinction is important for retention analysis because repeat
purchase behaviour must be measured at the underlying customer level,
not simply at the `customer_id` level.

The cardinality of both identifiers is therefore quantified before the
customer-level analytical table is constructed.

```python
customers = data["olist_customers_dataset.csv"]

customer_id_count = customers["customer_id"].nunique()
customer_unique_id_count = customers["customer_unique_id"].nunique()

print(f"Customer records: {len(customers):,}")
print(f"Unique customer_id: {customer_id_count:,}")
print(f"Unique customer_unique_id: {customer_unique_id_count:,}")
print(
    f"Customer records sharing an underlying customer identity: "
    f"{len(customers) - customer_unique_id_count:,}"
)
```

### Customer Identity Cardinality Result

The customer table contains 99,441 customer records and 99,441 unique
`customer_id` values, confirming that `customer_id` is unique within the
customer table.

There are 96,096 unique `customer_unique_id` values. Therefore, 3,345
customer records share an underlying customer identity.

For the retention analysis, `customer_unique_id` will be used as the
customer-level analytical grain so that repeat purchases are measured
across the same underlying customer rather than across separate
customer records.

###### 6.6 Composite Key — Order Items

Because an order can contain multiple item records, `order_id` alone
cannot uniquely identify a row in the order-items table.

The combination `(order_id, order_item_id)` is therefore validated as the
composite key for order-item records.

### 6.6 Composite Key — Order Items

Because an order can contain multiple item records, `order_id` alone
cannot uniquely identify a row in the order-items table.

The combination `(order_id, order_item_id)` is therefore validated as the
composite key for order-item records.

```python
# Validate the composite key for order items

order_items = data["olist_order_items_dataset.csv"]

duplicate_composite = order_items.duplicated(
    subset=["order_id", "order_item_id"]
).sum()

print(f"Duplicate (order_id, order_item_id) combinations: {duplicate_composite:,}")

if duplicate_composite == 0:
    print("PASS: (order_id, order_item_id) is a valid composite key.")
else:
    print("CHECK REQUIRED: duplicate composite keys found.")
```

### Composite Key Result

No duplicate `(order_id, order_item_id)` combinations were identified.
The composite key therefore uniquely identifies order-item records and
can be used to preserve item-level granularity before aggregation to the
order level.

## 7. Data Quality Assessment

This section assesses completeness, validity and potential anomalies in
the fields relevant to the proposed retention analysis.

The assessment distinguishes genuine missingness and structurally
non-applicable values from potential data-quality problems.

No records are removed and no values are imputed until a documented
treatment decision has been established.

### 7.1 Missing-Value Profile

```python
# Missing-value summary### 7.1 Missing-Value Profile

missing_summary = []

for filename, df in data.items():
    for column in df.columns:
        missing_count = df[column].isna().sum()

        if missing_count > 0:
            missing_summary.append({
                "file": filename,
                "field": column,
                "rows": len(df),
                "missing": missing_count,
                "missing_pct": round(
                    missing_count / len(df) * 100, 3
                )
            })

missing_summary = pd.DataFrame(missing_summary)

display(
    missing_summary.sort_values(
        ["file", "missing_pct"],
        ascending=[True, False]
    )
)
```

### Missing-Value Profile Result

Missing values are concentrated in review text, delivery-related dates and
selected product attributes.

Review comment title and message have substantial missingness. These
fields are not required for the quantitative retention analysis and review
text is therefore not imputed.

Missing delivery dates are primarily associated with orders that were not
successfully delivered and will be treated according to order status rather
than imputed.

Product attribute missingness is limited and will be assessed based on
whether the corresponding attributes are used in the analytical dataset.

No blanket missing-value imputation or row deletion is applied at this
stage.

```python
products = data["olist_products_dataset.csv"]

product_missing_cols = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
]

missing_product_attributes = products[
    product_missing_cols
].isna().all(axis=1)

print(
    "Products missing all four related attributes:",
    f"{missing_product_attributes.sum():,}"
)

print(
    "Products missing at least one of the four attributes:",
    f"{products[product_missing_cols].isna().any(axis=1).sum():,}"
)

display(
    products.loc[
        products[product_missing_cols].isna().any(axis=1),
        ["product_id"] + product_missing_cols
    ].head(10)
)
```

### Product Attribute Missingness Result

A total of 610 products are missing all four related product attributes:
`product_category_name`, `product_name_lenght`,
`product_description_lenght`, and `product_photos_qty`.

The same 610 products account for all products with at least one missing
value among these four fields, indicating that the attributes are missing
as a group rather than independently.

These missing values will not be imputed automatically. Product category
and descriptive attributes will only be used where the required source
information is available.

```python
incomplete_product_ids = set(
    products.loc[
        missing_product_attributes,
        "product_id"
    ]
)

order_items = data["olist_order_items_dataset.csv"]

incomplete_product_items = order_items[
    order_items["product_id"].isin(incomplete_product_ids)
]

print(
    "Incomplete products appearing in order items:",
    f"{incomplete_product_items['product_id'].nunique():,}"
)

print(
    "Order-item rows involving incomplete products:",
    f"{len(incomplete_product_items):,}"
)

print(
    "Orders involving incomplete products:",
    f"{incomplete_product_items['order_id'].nunique():,}"
)
```

### Transaction Coverage of Incomplete Products

All 610 products with missing product attributes appear in the order-items
table. They account for 1,603 order-item records across 1,451 orders.

These products therefore participate in actual transactions and cannot be
removed solely because their catalogue attributes are incomplete.

The corresponding transactions will be retained. Missing product
attributes will remain missing unless a specific analytical feature
requires an explicit treatment.

```python
physical_cols = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

physical_missing = products[
    physical_cols
].isna()

print(
    "Products missing at least one physical attribute:",
    f"{physical_missing.any(axis=1).sum():,}"
)

print(
    "Products missing all four physical attributes:",
    f"{physical_missing.all(axis=1).sum():,}"
)

display(
    products.loc[
        physical_missing.any(axis=1),
        ["product_id"] + physical_cols
    ]
)
```

### Physical Product Attribute Missingness Result

Only two products have missing physical attributes, and both are missing
all four fields: weight, length, height and width.

The missingness is therefore limited to two product records and does not
require broad imputation.

The affected products will be retained in the product catalogue. Their
transaction coverage will be checked before determining whether any
feature-specific treatment is required.

```python
physical_missing_product_ids = set(
    products.loc[
        physical_missing.any(axis=1),
        "product_id"
    ]
)

physical_missing_items = order_items[
    order_items["product_id"].isin(
        physical_missing_product_ids
    )
]

print(
    "Missing-physical-attribute products appearing in order items:",
    f"{physical_missing_items['product_id'].nunique():,}"
)

print(
    "Order-item rows involving these products:",
    f"{len(physical_missing_items):,}"
)

print(
    "Orders involving these products:",
    f"{physical_missing_items['order_id'].nunique():,}"
)
```

### Transaction Coverage of Missing Physical Attributes

Both products with missing physical attributes appear in the order-items
table, accounting for 18 order-item records across 16 orders.

The affected transactions will therefore be retained. If physical product
attributes are used as model features, the missing values will require an
explicit feature-level treatment rather than removal of the underlying
transactions.

### 7.2 Date and Delivery Quality Checks

The delivery date fields are interpreted in the context of order status.
Missing delivery dates may be structurally valid for orders that were
cancelled, unavailable or otherwise not delivered.

This check determines whether missing delivery dates are concentrated in
specific order statuses before defining the treatment rule.

```python
# missing delivery date by order status
orders = data["olist_orders_dataset.csv"]

delivery_missing_by_status = (
    orders.assign(
        delivery_date_missing=
        orders["order_delivered_customer_date"].isna()
    )
    .groupby("order_status")
    .agg(
        total_orders=("order_id", "size"),
        missing_delivery_date=("delivery_date_missing", "sum")
    )
    .reset_index()
)

delivery_missing_by_status["missing_pct"] = (
    delivery_missing_by_status["missing_delivery_date"]
    / delivery_missing_by_status["total_orders"]
    * 100
).round(2)

display(
    delivery_missing_by_status
    .sort_values("missing_delivery_date", ascending=False)
)
```

### Delivery-Date Missingness by Order Status

Missing `order_delivered_customer_date` is strongly associated with order
status.

All shipped, unavailable, invoiced, processing, created and approved
orders have a missing delivered-customer date. Canceled orders are also
almost entirely missing this date.

In contrast, 96,470 of 96,478 delivered orders have a delivered-customer
date.

This indicates that the missing delivery date is largely structurally
non-applicable rather than random missingness. Delivery dates will
therefore not be imputed. Delivery-duration and delay metrics will be
calculated only for orders with a valid delivered-customer date.

```python
# delivered orders with missing customer date
orders = data["olist_orders_dataset.csv"]

delivered_missing_date = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].isna())
]

display(delivered_missing_date)
```

```python
delivery_status_summary = (
    orders.groupby("order_status")
    .agg(
        orders=("order_id", "size"),
        delivered_date_present=(
            "order_delivered_customer_date",
            lambda x: x.notna().sum()
        )
    )
    .reset_index()
)

delivery_status_summary["delivery_date_pct"] = (
    delivery_status_summary["delivered_date_present"]
    / delivery_status_summary["orders"]
    * 100
).round(2)

display(delivery_status_summary)
```

### Delivered Orders with Missing Customer Delivery Date

Eight orders are marked as delivered but have no
`order_delivered_customer_date`.

Seven of these orders have a carrier delivery date, while one order has
both delivery-date fields missing. Since the actual customer delivery
date cannot be reliably reconstructed from the available fields, these
records will not be imputed.

The eight orders will remain in the raw data but will be excluded from
calculations that require the actual customer delivery date, such as
delivery duration and delivery-delay metrics.

```python
# delivery date chronology
orders_date_check = orders.copy()

date_columns = [
    "order_purchase_timestamp",
    "order_estimated_delivery_date",
    "order_delivered_customer_date"
]

for col in date_columns:
    orders_date_check[col] = pd.to_datetime(
        orders_date_check[col],
        errors="coerce"
    )

delivered_valid = orders_date_check[
    orders_date_check["order_status"].eq("delivered")
    & orders_date_check["order_delivered_customer_date"].notna()
    & orders_date_check["order_estimated_delivery_date"].notna()
    & orders_date_check["order_purchase_timestamp"].notna()
].copy()

delivered_before_purchase = (
    delivered_valid["order_delivered_customer_date"]
    < delivered_valid["order_purchase_timestamp"]
)

print(
    "Delivered orders with valid dates:",
    f"{len(delivered_valid):,}"
)

print(
    "Delivered date earlier than purchase date:",
    f"{delivered_before_purchase.sum():,}"
)

if delivered_before_purchase.sum() > 0:
    display(
        delivered_valid.loc[
            delivered_before_purchase,
            [
                "order_id",
                "order_purchase_timestamp",
                "order_delivered_customer_date",
                "order_estimated_delivery_date"
            ]
        ].head(20)
    )
```

### Delivery Date Chronology Result

All 96,470 delivered orders with a valid customer delivery date have a
delivery date on or after the purchase date.

No chronologically invalid delivery records were identified.

These records are therefore suitable for calculating delivery duration
and delivery-delay measures.

### Estimated Delivery Date Validity

The estimated delivery date is checked against the order purchase
timestamp to identify logically invalid delivery expectations before
calculating delivery delay.

```python
estimated_before_purchase = (
    delivered_valid["order_estimated_delivery_date"]
    < delivered_valid["order_purchase_timestamp"]
)

print(
    "Estimated delivery date earlier than purchase date:",
    f"{estimated_before_purchase.sum():,}"
)

if estimated_before_purchase.sum() > 0:
    display(
        delivered_valid.loc[
            estimated_before_purchase,
            [
                "order_id",
                "order_purchase_timestamp",
                "order_estimated_delivery_date",
                "order_delivered_customer_date"
            ]
        ].head(20)
    )
```

### Estimated Delivery Date Chronology Result

No orders have an estimated delivery date earlier than the purchase date.

The estimated delivery dates therefore pass the basic chronological
validity check and can be used to calculate delivery delay for orders
with a valid actual customer delivery date.

### Delivery Delay and Late-Delivery Profile

Delivery delay is calculated as the difference between the actual
customer delivery date and the estimated delivery date.

A positive delay indicates delivery after the estimated date. A zero or
negative value indicates delivery on or before the estimated date.

The late-delivery indicator is therefore:

    late = 1 if delivery_delay_days > 0
    late = 0 otherwise

```python
delivered_valid["delivery_delay_days"] = (
    delivered_valid["order_delivered_customer_date"]
    - delivered_valid["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)

delivered_valid["is_late"] = (
    delivered_valid["delivery_delay_days"] > 0
).astype(int)

print("Delivery delay summary:")
display(
    delivered_valid["delivery_delay_days"].describe()
)

print("\nLate-delivery classification:")
display(
    delivered_valid["is_late"]
    .value_counts()
    .rename_axis("is_late")
    .to_frame("orders")
)

print(
    f"\nLate-delivery rate: "
    f"{delivered_valid['is_late'].mean() * 100:.2f}%"
)
```

### Delivery Delay Result

Among the 96,470 delivered orders with valid dates, the median delivery
delay is -11.95 days, indicating that the typical order was delivered
ahead of the estimated delivery date.

A total of 7,826 orders were delivered after the estimated delivery date,
giving a late-delivery rate of 8.11%.

The `is_late` indicator is defined as delivery after the estimated
delivery date. Delivery delay is retained as a continuous variable for
analysis, while `is_late` provides a binary representation of delivery
performance.

Extreme delivery-delay values are investigated before deciding whether
they represent data errors or legitimate observations.

No outliers are removed or capped without evidence that the underlying
record is invalid.

### Investigation of Extreme Delivery Dates

Extreme delivery delays are not automatically treated as outliers.
Because the H1 variable is based on whether delivery occurred after the
estimated date, records are retained unless the underlying dates are
demonstrably invalid.

The concentration of very late deliveries around particular delivery
dates is investigated before defining any treatment.

```python
# Investigate concentration of delivery dates among extreme delays

extreme_late = delivered_valid[
    delivered_valid["delivery_delay_days"] >= 60
].copy()

print(f"Orders with delay >= 60 days: {len(extreme_late):,}")

print("\nMost common actual delivery dates among extreme delays:")
display(
    extreme_late["order_delivered_customer_date"]
    .dt.date
    .value_counts()
    .head(20)
)
```

```python
print("\nExtreme-delay orders by month of actual delivery:")

extreme_late["delivery_month"] = (
    extreme_late["order_delivered_customer_date"]
    .dt.to_period("M")
    .astype(str)
)

display(
    extreme_late["delivery_month"]
    .value_counts()
    .sort_index()
)
```

### Extreme Delivery Delay Assessment

A total of 84 delivered orders have a delivery delay of at least 60
days. These observations are concentrated in specific delivery periods,
with 25 occurring in September 2017, including 24 orders delivered on
2017-09-19.

The concentration suggests a potential data-quality or operational
anomaly, but the available date fields alone do not establish that these
records are erroneous.

The extreme observations will therefore be retained rather than removed
or winsorized. Their influence will be considered when interpreting
delivery-delay statistics and during subsequent model diagnostics.

### 7.2 Missing-Value Treatment Strategy

Missing values are handled according to their meaning and analytical
context rather than through blanket imputation.

- Review text fields are not imputed because review text is not used in
  the quantitative retention analysis.
- Missing customer delivery dates are not imputed. Delivery-duration and
  delivery-delay measures are calculated only where the actual customer
  delivery date is available.
- Product catalogue attributes with missing values are retained in the
  source data and are not used to justify removal of the associated
  transactions.
- The 610 products with missing category and descriptive attributes
  participate in valid transactions and are therefore retained.
- The two products with missing physical attributes are also retained.
  If these attributes are used as model features, feature-level missing
  value treatment will be applied at that stage.
- No blanket row deletion or global missing-value imputation is applied.

This approach preserves valid transactional observations while preventing
unsupported assumptions about missing values.
